### Key Findings

- Mumbai Indians have the highest number of match wins in the dataset.
- Choosing to field first was more common than choosing to bat first.
- Toss winner and match winner were the same in approximately 50.59% of matches.
- Wankhede Stadium hosted the highest number of matches in the dataset.
- 2013 had the highest number of matches among the seasons analysed.
- Matches were more often decided by wickets than by runs.
- Venue-wise result percentages varied, but venues with fewer matches need cautious interpretation.

```python
X = df[["season", "city", "match_type", "venue", "team1", "team2", "toss_winner", "toss_decision"]]
X.head(3)
```

```python
y = df["toss_match_result"]
```

```python
y.isnull().sum()
```

```python
categorical_cols = ["city", "match_type", "venue", "team1", "team2", "toss_winner", "toss_decision"]
```

```python
pd.set_option("display.max_columns", None)
X_encoded = pd.get_dummies(X,columns = categorical_cols,dtype=int)
X_encoded.head()
```

```python
X_encoded.shape
```

```python
#  data for train test and split
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_encoded, y, test_size=0.2, random_state=42
)
```

```python
X_train.shape,X_test.shape,y_train.shape,y_test.shape
```

```python
from sklearn.linear_model import LogisticRegression
#logistic-regression
model = LogisticRegression(max_iter=1000)
```

```python
# model tranning
model.fit(X_train, y_train)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
model.fit(X_train_scaled, y_train)
```

```python
y_pred = model.predict(X_test_scaled)
y_pred
```

```python
y_pred.shape
```

```python
pd.DataFrame({
    "Actual": y_test.values,
    "Predicted": y_pred
}).head(10)
```

```python
from sklearn.metrics import accuracy_score

accuracy_score(y_test, y_pred)
```

```python
from sklearn.metrics import confusion_matrix

confusion_matrix(y_test, y_pred)
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred))
```
